# Exploratory Data Analysis - SMS Spam Collection

```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd, numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from src.data_preprocessing import load_raw, clean_text
df = load_raw()
print(df.shape)
df.head()
```

```python
print('Missing values:\n', df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
print(df['label'].value_counts())
```

```python
os.makedirs('../reports/evaluation', exist_ok=True)
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='label')
plt.title('Spam vs Ham Distribution')
plt.savefig('../reports/evaluation/class_distribution.png')
plt.show()
```

```python
df['length'] = df['message'].astype(str).apply(len)
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='length', hue='label', bins=50, kde=True)
plt.title('Message Length Distribution')
plt.savefig('../reports/evaluation/message_length.png')
plt.show()
```

```python
df['cleaned'] = df['message'].apply(clean_text)
spam_words = ' '.join(df[df.label=='spam']['cleaned']).split()
ham_words = ' '.join(df[df.label=='ham']['cleaned']).split()
print('Top spam words:', Counter(spam_words).most_common(15))
print('Top ham words:', Counter(ham_words).most_common(15))
```

```python
fig, axes = plt.subplots(1,2, figsize=(14,5))
pd.Series(spam_words).value_counts().head(15).plot(kind='bar', ax=axes[0], title='Top Spam Words')
pd.Series(ham_words).value_counts().head(15).plot(kind='bar', ax=axes[1], title='Top Ham Words')
plt.tight_layout()
plt.savefig('../reports/evaluation/top_words.png')
plt.show()
```

```python
spam_counts = Counter(spam_words)
ham_counts = Counter(ham_words)
common = list(set(spam_counts) & set(ham_counts))[:20]
comp = pd.DataFrame({'spam':[spam_counts[w] for w in common], 'ham':[ham_counts[w] for w in common]}, index=common)
comp.plot(kind='bar', figsize=(12,5))
plt.title('Word Frequency Comparison')
plt.tight_layout()
plt.savefig('../reports/evaluation/word_freq_comparison.png')
plt.show()
```
